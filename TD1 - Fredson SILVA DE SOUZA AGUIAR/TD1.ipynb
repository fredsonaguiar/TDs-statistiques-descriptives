## Programmation Orientée Objet

```python
"""
Exemple de classe en Python
"""
class Moteur:

    # Constructeur
    def __init__(self, esn, panne=False):
        self.esn = esn
        self.panne = panne

    # Méthodes
    def dire_bonjour(self):
        print('Bonjour, mon numéro de série est ' + self.esn)

    def fonctionne(self):
        return not self.panne
```

```python
mot1 = Moteur('420912')
mot2 = Moteur(panne=True, esn='420913')

mot1.dire_bonjour()
print(mot1.fonctionne())

mot1.panne = True
print(mot1.fonctionne())

print('\n')
mot2.dire_bonjour()
print(mot2.fonctionne())
```

### Méthode récurcive vs Méthode itérative

```python
# Méthode récursive
def fact1(n):
    if n == 0:
        return 1
    else:
        return n*fact1(n-1)

# Méthode itérative
def fact2(n):
    s=1
    for i in range(1,n+1):
        s*=i
    return s

print(fact1(5))
print(fact2(5))
```

```python
"""
App.1 : Ecrire une fonction qui retourne le plus grand commun diviseur (pgcd) entre deux nombres a et b.
"""
```

```python
"""
App.1 : Correction
"""

#Méthode itérative
def pgcd1(a,b):
    while (a%b) != 0:
        s = a%b
        a = b
        b = s
    return b


#Méthode récurcive
def pgcd2(a,b):
    if a==b : return a
    if a<b : a,b = b,a
    return pgcd2(a-b,b)

a = 230
b = 25

print("pgcd1(a,b) = ", pgcd1(a,b))
print("pgcd2(a,b) = ", pgcd2(a,b))
```

$$ (AB)_{ij} = \sum_k A_{ik} B_{kj} $$

```python
"""
EXERCICE - Implémenter le produit matriciel de 2 matrices sous forme de listes de listes python
"""

def produit(A, B):
  rowsA = len(A); colsA = len(A[0])
  rowsB = len(B); colsB = len(B[0])

  prod = [[None for i in range(colsB)] for _ in range(rowsA)]

  for i in range(rowsA):
    for j in range(colsB):
      prod[i][j] = 0
      for k in range(colsA):
        prod[i][j] += A[i][k]*B[k][j]

  return prod
```

```python
# Vérification
assert(produit([[1, 2], [3, 4]], [[1, 2], [3, 4]]) == [[7, 10], [15, 22]])
```

```python
"""
EXERCICE - La méthode fillna
La méthode dropna permet d'imputer les valeurs manquantes (NaN). Lisez d'abord sa documentation.
1. Quelles sont les différentes stratégies de remplissage des valeurs manquantes ?
2. Imputez les valeurs manquantes de la colonne age du DF donné en exemple par :
    - 0
    - la dernière valeur précédente/suivante valide
    - la moyenne
    - la valeur la plus courante (mode)
3. Quel est le meilleur choix dans ce cas ? Et pour le cas d'une variable temporelle, par exemple la température 'EGT_SEL' de notre jeu de données ?
"""
exemple = pd.DataFrame({'nom': ['Alice', 'Bob', 'Charlie', 'David'], 'age': [24, np.nan, 99, 24]})
```

#### Q1 Ans:

The pandas.DataFrame.fillna method has many strategies and parameters describing different manners to fill missing values: the field 'value' allows one to parse a fixed value, or structured data to fill gaps; the field 'method' has options to fill a missing value by making use of the last/next valid value; also, the field named 'axis' receives the direction to wich consider when filling missing values.

#### Q2 Ans:

We do as follows for th different requirements:

```python
exemple_0 = exemple.fillna(value=0)
exemple_0
```

```python
exemple_1_ffill = exemple.ffill() # used ffill method because fillna(method="ffill") is depracated
exemple_1_ffill
```

```python
exemple_1_bfill = exemple.bfill() # used bfill method because fillna(method="bfill") is depracated
exemple_1_bfill
```

```python
example_2 = exemple.fillna(value=exemple["age"].mean())
example_2
```

```python
example_3 = exemple.fillna(value=exemple.mode().iloc[0])
example_3
```

#### Q3 Ans:

In fact, a choice of strategy for filling missing data depends strongly on the data and use. In this case, however, a good choice seems to be the average, since there is no relation between individuals that justify the use of a positional strategy, as 'bfill' and 'ffill'. Filling the missing field with the average value, on the other side, avoids changing the statistical distribution of the data in question.

On the other side, a temporal variable might (probably) has a temporal dependency, in this case, filling a missing field that varies over time, such as 'EGT_SEL' in our dataset, might require an strategy tha takes into consideration the time/sequential structure. Filling a missing temperature with the average observed may not be representative; instead, filling sich value with the previous/following or even the average of both may be a better option, especially if we assume the temperature transition to be smooth.
